Đánh giá và so sánh tỷ lệ báo động giả (False Positive Rate - FPR) và độ trễ (latency) của **4 cấu hình bộ test** khác nhau trên **100 bài nộp mô phỏng**.

```python
import os
import sys
from pathlib import Path

def _find_project_root(start_path: Path, max_up=6):
    p = start_path.resolve()
    for _ in range(max_up):
        if (p / "data").exists() and (p / "src").exists():
            return p
        if p.parent == p:
            break
        p = p.parent
    return start_path.resolve()

try:
    _cwd = Path.cwd()
    if str(_cwd).startswith("/content"):
        BASE = Path("/content/drive/MyDrive/Project")
    else:
        BASE = _find_project_root(_cwd)
except Exception:
    BASE = Path.cwd()

sys.path.insert(0, str(BASE / 'src'))
os.chdir(BASE)

import comparison_3sets
print("✓ Đang chạy chấm điểm so sánh 4 cấu hình kiểm thử...\n")
comparison_3sets.main()
```
