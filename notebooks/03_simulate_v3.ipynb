Tạo và đồng bộ hóa **100 bài nộp mô phỏng** chứa các lỗi thực tế (WA, RE, CE, TLE, MLE, hardcode, bảo mật) tương ứng với 100 bài toán MBPP mới.

```python
import os
import json
import sys
from pathlib import Path

def _find_project_root(start_path: Path, max_up=6):
    p = start_path.resolve()
    for _ in range(max_up):
        if (p / "data").exists() and (p / "src").exists():
            return p
        if p.parent == p:
            break
        p = p.parent
    return start_path.resolve()

try:
    _cwd = Path.cwd()
    if str(_cwd).startswith("/content"):
        BASE = Path("/content/drive/MyDrive/Project")
    else:
        BASE = _find_project_root(_cwd)
except Exception:
    BASE = Path.cwd()

sys.path.insert(0, str(BASE / 'src'))
os.chdir(BASE)

import simulate_v2
print("✓ Bắt đầu chạy sinh dữ liệu bài nộp học sinh (100 bài)...\n")
simulate_v2.generate()
```
