## Bước 2 — Kiểm tra tệp tin Tuần 4 (v3)

```python
from pathlib import Path

from pathlib import Path
BASE = next((p for p in [Path.cwd().resolve()] + list(Path.cwd().resolve().parents) if (p / 'data').exists() or (p / 'app.py').exists()), Path.cwd().resolve())

files_to_check = [
    BASE / "src" / "runner_v3.py",
    BASE / "data" / "processed" / "hidden_v2.json",
    BASE / "data" / "processed" / "submissions_50.json"
]

all_ok = True
for path in files_to_check:
    if path.exists():
        print(f"  ✓ {path.name} (Tồn tại)")
    else:
        print(f"  ✗ Thiếu: {path.name}")
        all_ok = False

if all_ok:
    print("\n✓ Đã đủ file cần thiết cho Tuần 4")
else:
    print("\n❌ Thiếu file — hãy kiểm tra lại thư mục dự án")
```
